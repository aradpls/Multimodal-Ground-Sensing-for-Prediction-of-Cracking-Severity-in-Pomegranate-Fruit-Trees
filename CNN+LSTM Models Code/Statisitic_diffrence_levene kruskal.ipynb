```python
#Import Libraries
from scipy.stats import levene, kruskal
import pandas as pd
```

```python
#AA comprison after model compariosn fole
df = pd.read_csv("/content/drive/MyDrive/Thesis/POM-IMG/row3and4_convnext_atten_16batch_30epochs/model_comparison_results.csv")

# Create correctness vectors
row3_correct = (df["Row3_Pred"] == df["True_Label"]).astype(int)
row4_correct = (df["Row4_Pred"] == df["True_Label"]).astype(int)

# Levene's Test for equal variance
levene_stat, levene_p = levene(row3_correct, row4_correct)

# Kruskal-Wallis for distribution
kruskal_stat, kruskal_p = kruskal(row3_correct, row4_correct)

print(f"Levene's Test → p = {levene_p:.4f} → {'Same variance' if levene_p > 0.05 else 'Different variance'}")
print(f"Kruskal-Wallis → p = {kruskal_p:.4f} → {'Same distribution' if kruskal_p > 0.05 else 'Different distribution'}")
```
